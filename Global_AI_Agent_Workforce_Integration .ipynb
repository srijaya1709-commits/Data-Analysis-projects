```python
import matplotlib.pyplot as plt

# Automatically pick the first two columns of your dataset:
x_col = df.columns[0]
y_col = df.columns[1]

# Matplotlib line chart
plt.figure(figsize=(10, 4))
plt.plot(
    df[x_col].head(25),
    df[y_col].head(25),
    marker='o',
    color='#1f77b4',
    linewidth=2,
)
plt.title(f'Matplotlib: {y_col} vs {x_col}', fontsize=13, fontweight='bold')
plt.xlabel(x_col)
plt.ylabel(y_col)
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
# seaborn line chart 
plt.figure(figsize=(10, 4))
sns.set_theme(style='whitegrid')
sns.lineplot(data=df.head(25), x=x_col, y=y_col, marker='o', color='coral')
plt.title(f'Seaborn: {y_col} vs {x_col}', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# ================= 1. MATPLOTLIB BAR CHART =================
sample_df = df.head(10)
plt.figure(figsize=(10, 5))
plt.bar(
    sample_df[x_col].astype(str),
    sample_df[y_col],
    color='#1f77b4',
    edgecolor='black',
)
plt.title(f'Matplotlib: {y_col} by {x_col}', fontsize=13, fontweight='bold')
plt.xlabel(x_col, fontsize=11)
plt.ylabel(y_col, fontsize=11)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Sample data definition
sample_df = pd.DataFrame({
    'Category': ['A', 'B', 'C', 'D', 'E'],
    'Value': [23, 45, 56, 78, 32]
})
x_col = 'Category'
y_col = 'Value'

# ================= 2. SEABORN BAR CHART =================
plt.figure(figsize=(10, 5))
sns.set_theme(style='whitegrid')
sns.barplot(data=sample_df, x=x_col, y=y_col, palette='viridis')
plt.title(f'Seaborn: {y_col} by {x_col}', fontsize=13, fontweight='bold')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# ================= MATPLOTLIB SCATTER PLOT =================
plt.figure(figsize=(10, 6))
plt.scatter(
    sample_df[x_col], 
    sample_df[y_col], 
    color='#1f77b4', 
    alpha=0.7, 
    edgecolors='black', 
    linewidth=0.5, 
    s=60
)
plt.title(f'Scatter Plot: {y_col} vs {x_col}', fontsize=14, fontweight='bold')
plt.xlabel(x_col, fontsize=12)
plt.ylabel(y_col, fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.set_theme(style='whitegrid')
sns.scatterplot(
    data=sample_df,
    x=x_col,
    y=y_col,
    palette='viridis',
    s=70,               # marker size
    alpha=0.8
)
plt.title(f'Seaborn: {y_col} vs {x_col}', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Sample 5x5 data matrix
np.random.seed(42)
data = np.random.randint(10, 100, size=(5, 5))
x_labels = [f'Col {i+1}' for i in range(5)]
y_labels = [f'Row {i+1}' for i in range(5)]
# ================= MATPLOTLIB HEATMAP =================
fig, ax = plt.subplots(figsize=(7, 6))
im = ax.imshow(data, cmap='viridis')
# Add ticks and labels
ax.set_xticks(range(len(x_labels)))
ax.set_yticks(range(len(y_labels)))
ax.set_xticklabels(x_labels)
ax.set_yticklabels(y_labels)
# Add colorbar
fig.colorbar(im, ax=ax, label='Values')
# Annotate each cell with its value
for i in range(len(y_labels)):
    for j in range(len(x_labels)):
        ax.text(j, i, str(data[i, j]), 
                ha="center", va="center", color="w", fontweight='bold')
```

```python
# Load built-in dataset and pivot into a matrix
flights = sns.load_dataset('flights')
pivot_table = flights.pivot(index='month', columns='year', values='passengers')
# ================= SEABORN HEATMAP =================
plt.figure(figsize=(10, 6))
sns.heatmap(
    pivot_table,
    annot=True,              # show numbers
    fmt='d',                 # integer format
    cmap='YlGnBu',           # Yellow-Green-Blue gradient
    linewidths=0.5
)
plt.title('Passengers Heatmap by Year & Month', fontsize=14, fontweight='bold')
plt.xlabel('Year', fontsize=12)
plt.ylabel('Month', fontsize=12)
plt.tight_layout()
plt.show()
```
