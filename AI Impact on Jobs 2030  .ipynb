```python
import matplotlib.pyplot as plt
import seaborn as sns

# Optional: Set a nice visual style for plots
sns.set_theme(style="whitegrid")
```

```python
import matplotlib.pyplot as plt

# Automatically pick the first two columns of your dataset:
x_col = df.columns[0]
y_col = df.columns[1]

# Matplotlib line chart
plt.figure(figsize=(10, 4))
plt.plot(
    df[x_col].head(25),
    df[y_col].head(25),
    marker='o',
    color='#1f77b4',
    linewidth=2,
)
plt.title(f'Matplotlib: {y_col} vs {x_col}', fontsize=13, fontweight='bold')
plt.xlabel(x_col)
plt.ylabel(y_col)
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
# seaborn line chart 
plt.figure(figsize=(10, 4))
sns.set_theme(style='whitegrid')
sns.lineplot(data=df.head(25), x=x_col, y=y_col, marker='o', color='coral')
plt.title(f'Seaborn: {y_col} vs {x_col}', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# ================= 1. MATPLOTLIB BAR CHART =================
sample_df = df.head(10)
plt.figure(figsize=(10, 5))
plt.bar(
    sample_df[x_col].astype(str),
    sample_df[y_col],
    color='#1f77b4',
    edgecolor='black',
)
plt.title(f'Matplotlib: {y_col} by {x_col}', fontsize=13, fontweight='bold')
plt.xlabel(x_col, fontsize=11)
plt.ylabel(y_col, fontsize=11)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python

# ================== 2. SEABORN BAR CHART ===================
plt.figure(figsize=(10, 5))
sns.set_theme(style='whitegrid')
sns.barplot(data=sample_df, x=x_col, y=y_col, palette='viridis')
plt.title(f'Seaborn: {y_col} by {x_col}', fontsize=13, fontweight='bold')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# ================= 1. MATPLOTLIB SCATTER PLOT =================
plt.figure(figsize=(9, 5))
plt.scatter(
    df[x_col],
    df[y_col],
    color='#1f77b4',
    alpha=0.7,
    edgecolors='black',
    s=50,  # point size
)
plt.title(
    f'Matplotlib Scatter: {y_col} vs {x_col}', fontsize=13, fontweight='bold'
)
plt.xlabel(x_col, fontsize=11)
plt.ylabel(y_col, fontsize=11)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# Automatically find numeric columns to plot
num_cols = df.select_dtypes(include=['number']).columns
plt.figure(figsize=(9, 5))
sns.set_theme(style='whitegrid')
# Scatter plot using the first two numeric columns
sns.scatterplot(
    data=df,
    x=num_cols[0],
    y=num_cols[1],
    color='#e74c3c',
    alpha=0.8,
    s=60,
)
plt.title(
    f'Seaborn Scatter: {num_cols[1]} vs {num_cols[0]}',
    fontsize=13,
    fontweight='bold',
)
plt.tight_layout()
plt.show()
```

```python
# 1. Compute the correlation matrix for all numeric columns
corr_matrix = df.select_dtypes(include=['number']).corr()
# ================= 2. MATPLOTLIB HEATMAP =================
plt.figure(figsize=(8, 6))
# Plot heatmap image
plt.imshow(corr_matrix, cmap='coolwarm', interpolation='nearest')
plt.colorbar()
# Add tick labels
cols = corr_matrix.columns
plt.xticks(range(len(cols)), cols, rotation=45, ha='right')
plt.yticks(range(len(cols)), cols)
plt.title('Correlation Heatmap (Matplotlib)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# ================== 1. SEABORN HEATMAP (Recommended) ==================
plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_matrix,
    annot=True,       # Show numbers inside each box
    cmap='coolwarm',  # Red = strong positive, Blue = strong negative
    fmt=".2f",        # 2 decimal places
    linewidths=0.5,
    cbar=True,
)
plt.title('Correlation Heatmap (Seaborn)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```
