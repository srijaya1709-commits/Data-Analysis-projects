Step 1: Import NumPy and Pandas & Load Data

```python
import numpy as np
import pandas as pd

# Load dataset
df = pd.read_csv("d:\REZILYENS Intern\cybersecurity dataset\cybersecurity_attacks.csv")
df.head()
```

Step 2: Check Missing Values

```python
# Check count of missing values per column
df.isna().sum()
```

Step 3: Check Null Values

```python
# Check total null values in the entire dataset
print("Total Null Values:", df.isnull().sum().sum())
```

Step 4: Check Duplicates

```python
# Check total duplicate rows
print("Total Duplicate Rows:", df.duplicated().sum())
```

Step 5: Drop (Duplicates & Nulls)

```python
# Drop duplicates
df = df.drop_duplicates()

# Drop rows with null / missing values
df = df.dropna()
```

Step 6: Clean the Data (Verify Result)

```python
# Verify everything is clean
print("Clean Data Shape:", df.shape)
print("Remaining Nulls:", df.isnull().sum().sum())
print("Remaining Duplicates:", df.duplicated().sum())

# View cleaned data
df.head()
```

 import matplotlib and seaborn:

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Optional: Set a nice visual style for plots
sns.set_theme(style="whitegrid")
```

1. Line chart 
Matplotlib

```python
col1 = df.select_dtypes(include='number').columns[0]
col2 = df.select_dtypes(include='number').columns[1]

plt.figure(figsize=(8, 4))
plt.plot(df[col1].head(30), df[col2].head(30), color='blue', marker='o')
plt.title(f"Line Plot (Matplotlib): {col1} vs {col2}")
plt.xlabel(col1)
plt.ylabel(col2)
plt.grid(True)
plt.show()
```

Seaborn

```python
col1 = df.select_dtypes(include='number').columns[0]
col2 = df.select_dtypes(include='number').columns[1]

plt.figure(figsize=(8, 4))
sns.lineplot(data=df.head(30), x=col1, y=col2, marker='o', color='blue')
plt.title(f"Line Plot (Seaborn): {col1} vs {col2}")
plt.show()
```

2. Bar Chart
Matplotlib

```python
col1 = df.columns[0]
col2 = df.select_dtypes(include='number').columns[0]

plt.figure(figsize=(8, 4))
# Plots first 10 rows so bars are clean and readable
plt.bar(df[col1].head(10).astype(str), df[col2].head(10), color='teal')
plt.title(f"Bar Plot (Matplotlib): {col1} vs {col2}")
plt.xlabel(col1)
plt.ylabel(col2)
plt.xticks(rotation=45)
plt.show()
```

Seaborn:

```python
col1 = df.columns[0]
col2 = df.select_dtypes(include='number').columns[0]

plt.figure(figsize=(8, 4))
sns.barplot(data=df.head(10), x=col1, y=col2, palette='viridis')
plt.title(f"Bar Plot (Seaborn): {col1} vs {col2}")
plt.xticks(rotation=45)
plt.show()
```

3. Scatter Plot
Matplotlib:

```python
col1 = df.select_dtypes(include='number').columns[0]
col2 = df.select_dtypes(include='number').columns[1]

plt.figure(figsize=(8, 4))
plt.scatter(df[col1], df[col2], color='coral', alpha=0.7)
plt.title(f"Scatter Plot (Matplotlib): {col1} vs {col2}")
plt.xlabel(col1)
plt.ylabel(col2)
plt.grid(True)
plt.show()
```

Seaborn:

```python
col1 = df.select_dtypes(include='number').columns[0]
col2 = df.select_dtypes(include='number').columns[1]

plt.figure(figsize=(8, 4))
sns.scatterplot(data=df, x=col1, y=col2, color='coral')
plt.title(f"Scatter Plot (Seaborn): {col1} vs {col2}")
plt.show()
```

4. Heatmap (Correlation Matrix)
Matplotlib

```python
# 1. Filter numeric columns first, then calculate correlation
corr = df.select_dtypes(include='number').corr()

# 2. Plot heatmap
plt.figure(figsize=(8, 6))
plt.imshow(corr, cmap='coolwarm', interpolation='none')
plt.colorbar()
plt.xticks(range(len(corr)), corr.columns, rotation=45)
plt.yticks(range(len(corr)), corr.columns)
plt.title("Heatmap (Matplotlib)")
plt.show()
```

seaborn 

```python
corr = df.select_dtypes(include='number').corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title("Heatmap (Seaborn)")
plt.show()
```
